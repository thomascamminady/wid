# German wealth distribution (WID.world)

Loads the tidy g-percentile data through `wid.io` and draws the charts from `wid.plotting`.


```python
from pathlib import Path

import polars as pl

from wid.io import DEFAULT_CSV, load_cumulative, load_gpercentiles
from wid.plotting import plot_donuts_zoom

# Paths in wid are relative to the repo root; the notebook runs from notebooks/.
REPO_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
CSV_PATH = REPO_ROOT / DEFAULT_CSV
YEAR = 2024
```

## All years, all 127 g-percentile bins


```python
df = load_gpercentiles(CSV_PATH)
df
```

## Top shares in one year

`load_cumulative` gives the share held by the richest x% of adults.


```python
cumulative = load_cumulative(CSV_PATH, YEAR)
cumulative.filter(pl.col("top_pct").is_in([50.0, 10.0, 1.0, 0.1, 0.01, 0.001]))
```

## Charts


```python
plot_donuts_zoom(
    cumulative, YEAR, REPO_ROOT / "output" / "germany_wealth_donuts_zoom.png"
)
```
